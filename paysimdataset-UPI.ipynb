```python
from google.colab import drive
drive.mount('/content/drive')
import os
import pandas as pd
import numpy as np
import gc
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D, Bidirectional, GRU, Dense,
    Dropout, Flatten, Lambda, GlobalAveragePooling1D,
    MultiHeadAttention, LayerNormalization, Add, Permute
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, auc
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

data_path = '/content/drive/MyDrive/project_ashwini_paper_2/paysimdataset.csv'
df = pd.read_csv(data_path)
print("Original shape:", df.shape)

def reduce_mem_usage(df):
    for col in df.columns:
        col_type = df[col].dtype
        if col_type != object:
            c_min = df[col].min()
            c_max = df[col].max()
            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                else:
                    df[col] = df[col].astype(np.int64)
            else:
                if c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float64)
        else:
            df[col] = df[col].astype('category')
    return df

df = reduce_mem_usage(df)

feature_cols = ['step', 'type', 'amount', 'oldbalanceOrg', 'newbalanceOrig',
                'oldbalanceDest', 'newbalanceDest']
target_col = 'isFraud'
df = df[feature_cols + [target_col]]

df['type'] = df['type'].astype(str)
le_type = LabelEncoder()
df['type'] = le_type.fit_transform(df['type'])

fraud = df[df[target_col] == 1]
non_fraud = df[df[target_col] == 0]

n_samples = 15000
if len(fraud) > n_samples:
    fraud = fraud.sample(n=n_samples, random_state=42)
if len(non_fraud) > n_samples:
    non_fraud = non_fraud.sample(n=n_samples, random_state=42)

df_balanced = pd.concat([fraud, non_fraud]).sample(frac=1, random_state=42).reset_index(drop=True)
print("Balanced shape:", df_balanced.shape)

X = df_balanced[feature_cols].values
y = df_balanced[target_col].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = X_scaled.reshape((X_scaled.shape[0], 1, X_scaled.shape[1]))
y_cat = to_categorical(y)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_cat, test_size=0.3, random_state=42)
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)

def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0):
    x = LayerNormalization(epsilon=1e-6)(inputs)
    x = MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(x, x)
    x = Add()([inputs, x])
    res = x
    x = LayerNormalization(epsilon=1e-6)(x)
    x = Dense(ff_dim, activation="relu")(x)
    x = Dropout(dropout)(x)
    x = Dense(inputs.shape[-1])(x)
    x = Add()([res, x])
    return x

@tf.custom_gradient
def gradient_reverse_op(x, lambda_):
    def grad(dy):
        return -lambda_ * dy, None
    return x, grad

class GradientReversalLayer(tf.keras.layers.Layer):
    def __init__(self, lambda_=1.0, **kwargs):
        super(GradientReversalLayer, self).__init__(**kwargs)
        self.lambda_ = lambda_

    def call(self, x, training=None):
        if training:
            return gradient_reverse_op(x, self.lambda_)
        return x

    def compute_output_shape(self, input_shape):
        return input_shape

    def get_config(self):
        config = super().get_config()
        config.update({'lambda_': self.lambda_})
        return config

def build_cnn_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Conv1D(32, 3, activation='relu', padding='same')(permuted)
    x = MaxPooling1D(2)(x)
    x = Conv1D(64, 3, activation='relu', padding='same')(x)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_bigru_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Bidirectional(GRU(64, return_sequences=True))(permuted)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_transformer_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Dense(64, activation='relu')(permuted)
    x = transformer_encoder(x, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_proposed_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    cnn = Conv1D(32, 3, activation='relu', padding='same')(permuted)
    cnn = MaxPooling1D(2)(cnn)
    cnn = Conv1D(64, 3, activation='relu', padding='same')(cnn)
    cnn = GlobalAveragePooling1D()(cnn)
    cnn = Dense(64, activation='relu')(cnn)
    gru = Bidirectional(GRU(64, return_sequences=True))(permuted)
    gru = GlobalAveragePooling1D()(gru)
    gru = Dense(64, activation='relu')(gru)
    trans = Dense(64, activation='relu')(permuted)
    trans = transformer_encoder(trans, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    trans = GlobalAveragePooling1D()(trans)
    trans = Dense(64, activation='relu')(trans)
    fused = Lambda(lambda t: tf.stack(t, axis=1), output_shape=(3, 64))([cnn, gru, trans])
    attn = MultiHeadAttention(num_heads=4, key_dim=32)(fused, fused)
    attn = Flatten()(attn)
    x = Dense(64, activation='relu')(attn)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_adversarial_game_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)

    cnn_branch = Conv1D(32, 3, activation='relu', padding='same')(permuted)
    cnn_branch = MaxPooling1D(2)(cnn_branch)
    cnn_branch = Conv1D(64, 3, activation='relu', padding='same')(cnn_branch)
    cnn_branch = GlobalAveragePooling1D()(cnn_branch)
    cnn_branch = Dense(64, activation='relu')(cnn_branch)

    gru_branch = Bidirectional(GRU(64, return_sequences=True))(permuted)
    gru_branch = GlobalAveragePooling1D()(gru_branch)
    gru_branch = Dense(64, activation='relu')(gru_branch)

    transformer_proj = Dense(64, activation='relu')(permuted)
    transformer_branch = transformer_encoder(transformer_proj, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    transformer_branch = GlobalAveragePooling1D()(transformer_branch)
    transformer_branch = Dense(64, activation='relu')(transformer_branch)

    fused = Lambda(lambda tensors: tf.stack(tensors, axis=1),
                   output_shape=(3, 64))([cnn_branch, gru_branch, transformer_branch])
    attention = MultiHeadAttention(num_heads=4, key_dim=32)(fused, fused)
    attention = Flatten()(attention)
    fusion_features = Dense(64, activation='relu', name='fusion_features')(attention)

    x = Dropout(0.5)(fusion_features)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    class_outputs = Dense(num_classes, activation='softmax', name='classifier')(x)

    adv_branch = GradientReversalLayer(lambda_=0.1, name='grad_reverse')(fusion_features)
    adv_branch = Dense(32, activation='relu')(adv_branch)
    adv_branch = Dense(32, activation='relu')(adv_branch)
    disc_outputs = Dense(1, activation='sigmoid', name='discriminator')(adv_branch)

    model = Model(inputs, [class_outputs, disc_outputs])
    return model

class AdversarialGameModel(Model):
    def __init__(self, base_model, epsilon=0.01, **kwargs):
        super(AdversarialGameModel, self).__init__(**kwargs)
        self.base_model = base_model
        self.epsilon = epsilon
        self.disc_loss_tracker = tf.keras.metrics.Mean(name='disc_loss')
        self.class_loss_tracker = tf.keras.metrics.Mean(name='class_loss')
        self.total_loss_tracker = tf.keras.metrics.Mean(name='loss')

    def compile(self, optimizer, class_loss_fn, disc_loss_fn):
        super(AdversarialGameModel, self).compile()
        self.optimizer = optimizer
        self.class_loss_fn = class_loss_fn
        self.disc_loss_fn = disc_loss_fn

    def generate_adversarial(self, x, y_true):
        x = tf.cast(x, tf.float32)
        with tf.GradientTape() as tape:
            tape.watch(x)
            preds, _ = self.base_model(x, training=True)
            loss = self.class_loss_fn(y_true, preds)
        grad = tape.gradient(loss, x)
        if grad is None:
            return x
        signed_grad = tf.sign(grad)
        x_adv = x + self.epsilon * signed_grad
        x_adv = tf.clip_by_value(x_adv, tf.reduce_min(x), tf.reduce_max(x))
        return x_adv

    def train_step(self, data):
        x, y_true = data
        batch_size = tf.shape(x)[0]

        x_adv = self.generate_adversarial(x, y_true)
        clean_labels = tf.ones((batch_size, 1))
        adv_labels = tf.zeros((batch_size, 1))
        combined_x = tf.concat([x, x_adv], axis=0)
        combined_disc_labels = tf.concat([clean_labels, adv_labels], axis=0)

        with tf.GradientTape() as tape:
            class_preds, disc_preds = self.base_model(combined_x, training=True)
            class_preds_clean = class_preds[:batch_size]
            class_preds_adv = class_preds[batch_size:]
            class_loss_clean = self.class_loss_fn(y_true, class_preds_clean)
            class_loss_adv = self.class_loss_fn(y_true, class_preds_adv)
            class_loss = class_loss_clean + 0.5 * class_loss_adv
            disc_loss = self.disc_loss_fn(combined_disc_labels, disc_preds)
            total_loss = class_loss + disc_loss

        trainable_vars = self.base_model.trainable_variables
        grads = tape.gradient(total_loss, trainable_vars)
        self.optimizer.apply_gradients(zip(grads, trainable_vars))

        self.total_loss_tracker.update_state(total_loss)
        self.class_loss_tracker.update_state(class_loss)
        self.disc_loss_tracker.update_state(disc_loss)
        self.compiled_metrics.update_state(y_true, class_preds_clean)

        logs = {m.name: m.result() for m in self.metrics}
        logs['class_loss'] = self.class_loss_tracker.result()
        logs['disc_loss'] = self.disc_loss_tracker.result()
        logs['loss'] = self.total_loss_tracker.result()
        return logs

    def test_step(self, data):
        x, y_true = data
        class_preds, disc_preds = self.base_model(x, training=False)
        class_loss = self.class_loss_fn(y_true, class_preds)
        clean_labels = tf.ones((tf.shape(x)[0], 1))
        disc_loss = self.disc_loss_fn(clean_labels, disc_preds)
        total_loss = class_loss + disc_loss

        self.total_loss_tracker.update_state(total_loss)
        self.class_loss_tracker.update_state(class_loss)
        self.disc_loss_tracker.update_state(disc_loss)
        self.compiled_metrics.update_state(y_true, class_preds)

        logs = {m.name: m.result() for m in self.metrics}
        logs['class_loss'] = self.class_loss_tracker.result()
        logs['disc_loss'] = self.disc_loss_tracker.result()
        logs['loss'] = self.total_loss_tracker.result()
        return logs

    def call(self, inputs, training=False):
        class_preds, _ = self.base_model(inputs, training=training)
        return class_preds

class StopAtAccuracy(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        if logs.get('val_accuracy') is not None and logs.get('val_accuracy') >= 0.98:
            self.model.stop_training = True
            print(f"\nReached 98% validation accuracy, stopping at epoch {epoch+1}")

def train_and_evaluate(model, name, X_train, y_train, X_test, y_test, target_names):
    early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
    stop_acc = StopAtAccuracy()
    history = model.fit(X_train, y_train, validation_data=(X_test, y_test),
                        epochs=20, batch_size=64, callbacks=[early_stop, stop_acc], verbose=1)
    plt.plot(history.history['accuracy'], label='Train')
    plt.plot(history.history['val_accuracy'], label='Val')
    plt.title(f'{name} Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend()
    plt.show()
    plt.plot(history.history['loss'], label='Train')
    plt.plot(history.history['val_loss'], label='Val')
    plt.title(f'{name} Loss')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend()
    plt.show()
    y_probs = model.predict(X_test)
    y_pred = np.argmax(y_probs, axis=1)
    y_true = np.argmax(y_test, axis=1)
    print(f"\n{name} Classification Report:")
    report = classification_report(y_true, y_pred, target_names=target_names, zero_division=0, output_dict=True)
    print(classification_report(y_true, y_pred, target_names=target_names, zero_division=0))
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
    plt.title(f'{name} Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.show()
    plt.figure(figsize=(6,5))
    for i in range(len(target_names)):
        fpr, tpr, _ = roc_curve(y_test[:, i], y_probs[:, i])
        plt.plot(fpr, tpr, label=f'{target_names[i]} (AUC={auc(fpr, tpr):.2f})')
    plt.plot([0,1],[0,1],'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{name} ROC Curves')
    plt.legend()
    plt.show()
    auc_score = roc_auc_score(y_test, y_probs, multi_class='ovr')
    print(f"{name} AUC (OvR): {auc_score:.4f}")
    return report, auc_score

target_names = ['Non-Fraud', 'Fraud']

cnn_model = build_cnn_model(X_train.shape[1:], y_train.shape[1])
bigru_model = build_bigru_model(X_train.shape[1:], y_train.shape[1])
trans_model = build_transformer_model(X_train.shape[1:], y_train.shape[1])
proposed_model = build_proposed_model(X_train.shape[1:], y_train.shape[1])

base_adver_game_model = build_adversarial_game_model(X_train.shape[1:], y_train.shape[1])
adver_game_model = AdversarialGameModel(base_adver_game_model, epsilon=0.01)
adver_game_model.compile(
    optimizer=Adam(0.0005),
    class_loss_fn=tf.keras.losses.CategoricalCrossentropy(),
    disc_loss_fn=tf.keras.losses.BinaryCrossentropy()
)

print("\nTraining CNN...")
cnn_rep, cnn_auc = train_and_evaluate(cnn_model, "CNN", X_train, y_train, X_test, y_test, target_names)
print("\nTraining BiGRU...")
bigru_rep, bigru_auc = train_and_evaluate(bigru_model, "BiGRU", X_train, y_train, X_test, y_test, target_names)
print("\nTraining Transformer...")
trans_rep, trans_auc = train_and_evaluate(trans_model, "Transformer", X_train, y_train, X_test, y_test, target_names)
print("\nTraining Proposed...")
prop_rep, prop_auc = train_and_evaluate(proposed_model, "Proposed", X_train, y_train, X_test, y_test, target_names)

print("\nTraining Proposed with Adversarial Game Theory...")
adver_game_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=64,
    callbacks=[EarlyStopping(monitor='val_loss', mode='min', patience=5, restore_best_weights=True), StopAtAccuracy()],
    verbose=1
)

class_preds_adver = adver_game_model.predict(X_test)
y_pred_adver = np.argmax(class_preds_adver, axis=1)
y_true_adver = np.argmax(y_test, axis=1)

print("\nProposed + Adversarial Game Classification Report:")
adver_report = classification_report(y_true_adver, y_pred_adver, target_names=target_names, zero_division=0, output_dict=True)
print(classification_report(y_true_adver, y_pred_adver, target_names=target_names, zero_division=0))

cm_adver = confusion_matrix(y_true_adver, y_pred_adver)
plt.figure(figsize=(6,5))
sns.heatmap(cm_adver, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.title('Proposed + Adversarial Game Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()

plt.figure(figsize=(6,5))
for i in range(len(target_names)):
    fpr, tpr, _ = roc_curve(y_test[:, i], class_preds_adver[:, i])
    plt.plot(fpr, tpr, label=f'{target_names[i]} (AUC={auc(fpr, tpr):.2f})')
plt.plot([0,1],[0,1],'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Proposed + Adversarial Game ROC Curves')
plt.legend()
plt.show()

adver_auc = roc_auc_score(y_test, class_preds_adver, multi_class='ovr')
print(f"Proposed + Adversarial Game AUC (OvR): {adver_auc:.4f}")

models = ['CNN', 'BiGRU', 'Transformer', 'Proposed', 'Prop+AdvGame']
acc = [cnn_rep['accuracy'], bigru_rep['accuracy'], trans_rep['accuracy'], prop_rep['accuracy'], adver_report['accuracy']]
prec = [cnn_rep['macro avg']['precision'], bigru_rep['macro avg']['precision'],
        trans_rep['macro avg']['precision'], prop_rep['macro avg']['precision'], adver_report['macro avg']['precision']]
rec = [cnn_rep['macro avg']['recall'], bigru_rep['macro avg']['recall'],
       trans_rep['macro avg']['recall'], prop_rep['macro avg']['recall'], adver_report['macro avg']['recall']]
f1 = [cnn_rep['macro avg']['f1-score'], bigru_rep['macro avg']['f1-score'],
      trans_rep['macro avg']['f1-score'], prop_rep['macro avg']['f1-score'], adver_report['macro avg']['f1-score']]
auc_s = [cnn_auc, bigru_auc, trans_auc, prop_auc, adver_auc]

x = np.arange(len(models))
width = 0.15
fig, ax = plt.subplots(figsize=(14,6))
ax.bar(x - 2*width, acc, width, label='Accuracy')
ax.bar(x - width, prec, width, label='Precision')
ax.bar(x, rec, width, label='Recall')
ax.bar(x + width, f1, width, label='F1-score')
ax.bar(x + 2*width, auc_s, width, label='AUC')
ax.set_xlabel('Models')
ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison (PaySim)')
ax.set_xticks(x)
ax.set_xticklabels(models)
ax.legend()
plt.tight_layout()
plt.show()
```
