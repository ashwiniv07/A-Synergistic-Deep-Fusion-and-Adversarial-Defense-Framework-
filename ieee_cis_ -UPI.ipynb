```python
from google.colab import drive
drive.mount('/content/drive')
path = "/content/drive/MyDrive/project_ashwini_paper_2/ieee-fraud-detection"
import os
import pandas as pd
import numpy as np
import gc
from sklearn.preprocessing import LabelEncoder, StandardScaler
from collections import Counter
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import pairwise_distances
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D, Bidirectional, GRU, Dense,
    Dropout, Flatten, Attention, Permute, Multiply, Lambda, GlobalAveragePooling1D,
    MultiHeadAttention, LayerNormalization, Add
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow.keras.backend as K
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, f1_score, roc_curve, auc
import seaborn as sns

print("Path to dataset files:", path)

train_trans = pd.read_csv(os.path.join(path, 'train_transaction.csv'))
train_id = pd.read_csv(os.path.join(path, 'train_identity.csv'))

def reduce_mem_usage(df):
    for col in df.columns:
        col_type = df[col].dtype
        if col_type != object:
            c_min = df[col].min()
            c_max = df[col].max()
            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                else:
                    df[col] = df[col].astype(np.int64)
            else:
                if c_min > np.finfo(np.float16).min and c_max < np.finfo(np.float16).max:
                    df[col] = df[col].astype(np.float16)
                elif c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float64)
        else:
            df[col] = df[col].astype('category')
    return df

train_trans = reduce_mem_usage(train_trans)
train_id = reduce_mem_usage(train_id)
train = train_trans.merge(train_id, on='TransactionID', how='left')
del train_trans, train_id
gc.collect()

cat_cols = ['ProductCD', 'card1', 'card2', 'card3', 'card4', 'card5', 'card6',
            'addr1', 'addr2', 'P_emaildomain', 'R_emaildomain', 'M1','M2','M3',
            'M4','M5','M6','M7','M8','M9','DeviceType','DeviceInfo',
            'id_12','id_13','id_14','id_15','id_16','id_17','id_18','id_19',
            'id_20','id_21','id_22','id_23','id_24','id_25','id_26','id_27',
            'id_28','id_29','id_30','id_31','id_32','id_33','id_34','id_35',
            'id_36','id_37','id_38']
num_cols = ['TransactionAmt', 'TransactionDT']
target = 'isFraud'

for col in cat_cols:
    if col in train.columns:
        train[col] = train[col].astype(str).fillna('NaN')
        le = LabelEncoder()
        le.fit(train[col])
        train[col] = le.transform(train[col]).astype(np.int32)

for col in num_cols:
    if col in train.columns:
        train[col] = train[col].fillna(0).astype(np.float32)

train[target] = train[target].astype(np.float32)

features = cat_cols + num_cols
available_features = [f for f in features if f in train.columns]
X_full = train[available_features].values
y_full = train[target].values

del train
gc.collect()

class OptimizedNeuroSyncSampler:
    def __init__(self,
                 neural_depth=2,
                 evolutionary_generations=5,
                 quantum_entanglement_factor=0.7,
                 balance_strategy='adaptive',
                 target_samples_per_class=10000,
                 random_state=42):
        self.neural_depth = neural_depth
        self.evolutionary_generations = evolutionary_generations
        self.quantum_factor = quantum_entanglement_factor
        self.balance_strategy = balance_strategy
        self.target_samples = target_samples_per_class
        self.random_state = random_state
        np.random.seed(random_state)
        self.standard_scaler = StandardScaler()
        self.feature_importance_scores_ = None

    def _neural_density_estimation(self, X, y):
        unique_classes = np.unique(y)
        density_scores = {}
        for cls in unique_classes:
            X_cls = X[y == cls]
            if len(X_cls) == 0:
                continue
            if len(X_cls) > 500000:
                sample_indices = np.random.choice(len(X_cls), 10000, replace=False)
                X_cls = X_cls[sample_indices]
            density_layers = []
            for depth in range(self.neural_depth):
                bandwidth = 0.5 + (0.3 * depth / self.neural_depth)
                chunk_size = min(1000, len(X_cls))
                layer_density = np.zeros(len(X_cls))
                for i in range(0, len(X_cls), chunk_size):
                    end_idx = min(i + chunk_size, len(X_cls))
                    chunk = X_cls[i:end_idx]
                    distances = pairwise_distances(chunk, X_cls, metric='euclidean')
                    kernel_vals = np.exp(-distances**2 / (2 * bandwidth**2))
                    chunk_density = np.mean(kernel_vals, axis=1)
                    layer_density[i:end_idx] = chunk_density
                density_layers.append(layer_density)
                del distances, kernel_vals, chunk_density
                gc.collect()
            fused_density = np.mean(density_layers, axis=0)
            density_scores[cls] = fused_density
        return density_scores

    def _evolutionary_undersampling(self, X_maj, X_min, fitness_scores):
        n_target = min(self.target_samples, len(X_min))
        if len(X_maj) > 20000:
            sample_indices = np.random.choice(len(X_maj), 20000, replace=False)
            population = X_maj[sample_indices]
            fitness = fitness_scores[sample_indices]
        else:
            population = X_maj.copy()
            fitness = fitness_scores.copy()
        for generation in range(self.evolutionary_generations):
            tournament_size = min(3, len(population))
            selected_indices = []
            while len(selected_indices) < n_target:
                if len(population) == 0:
                    break
                contestants = np.random.choice(len(population), min(tournament_size, len(population)), replace=False)
                winner = contestants[np.argmax(fitness[contestants])]
                selected_indices.append(winner)
            if len(selected_indices) == 0:
                break
            population = population[selected_indices]
            fitness = fitness[selected_indices]
            if generation < self.evolutionary_generations - 1:
                mutation = np.random.normal(0, 0.05, size=population.shape)
                population = population + mutation
        return population[:n_target]

    def _quantum_noise_detection(self, X, y):
        unique_classes = np.unique(y)
        noise_mask = np.zeros(len(X), dtype=bool)
        for cls in unique_classes:
            X_cls = X[y == cls]
            if len(X_cls) < 10:
                continue
            centroid = np.mean(X_cls, axis=0)
            distances = np.linalg.norm(X_cls - centroid, axis=1)
            threshold = np.percentile(distances, 95)
            class_indices = np.where(y == cls)[0]
            class_noise_mask = distances > threshold
            noise_mask[class_indices] = class_noise_mask
            del distances
            gc.collect()
        return noise_mask

    def _calculate_feature_importance(self, X, y):
        unique_classes = np.unique(y)
        importance_scores = np.zeros(X.shape[1])
        if len(X) > 10000:
            sample_indices = np.random.choice(len(X), 10000, replace=False)
            X_sample = X[sample_indices]
            y_sample = y[sample_indices]
        else:
            X_sample = X
            y_sample = y
        for feature_idx in range(X_sample.shape[1]):
            feature_vals = X_sample[:, feature_idx]
            class_means = []
            for cls in unique_classes:
                class_vals = feature_vals[y_sample == cls]
                if len(class_vals) > 0:
                    class_means.append(np.mean(class_vals))
            if len(class_means) > 1:
                between_class_var = np.var(class_means)
            else:
                between_class_var = 0
            if between_class_var > 0:
                importance_scores[feature_idx] = between_class_var
            else:
                importance_scores[feature_idx] = 0
        if np.sum(importance_scores) > 0:
            importance_scores = importance_scores / np.sum(importance_scores)
        self.feature_importance_scores_ = importance_scores
        return importance_scores

    def fit_resample(self, X, y):
        X = np.array(X)
        y = np.array(y)
        gc.collect()
        feature_importance = self._calculate_feature_importance(X, y)
        X_weighted = X * feature_importance
        X_scaled = self.standard_scaler.fit_transform(X_weighted)
        class_counts = Counter(y)
        target_size = self.target_samples
        unique_classes = np.unique(y)
        X_resampled_list = []
        y_resampled_list = []
        for cls in unique_classes:
            X_cls = X_scaled[y == cls]
            y_cls = y[y == cls]
            n_current = len(X_cls)
            if n_current == 0:
                continue
            if n_current < target_size:
                n_needed = target_size - n_current
                if n_needed > 0:
                    oversample_indices = np.random.choice(n_current, n_needed, replace=True)
                    synthetic_samples = X_cls[oversample_indices] + np.random.normal(0, 0.01, X_cls[oversample_indices].shape)
                    X_oversampled = np.vstack([X_cls, synthetic_samples])
                    y_oversampled = np.concatenate([y_cls, [cls] * n_needed])
                else:
                    X_oversampled = X_cls
                    y_oversampled = y_cls
                X_resampled_list.append(X_oversampled)
                y_resampled_list.append(y_oversampled)
            elif n_current > target_size:
                selected_indices = np.random.choice(n_current, target_size, replace=False)
                X_resampled_list.append(X_cls[selected_indices])
                y_resampled_list.append(np.array([cls] * target_size))
            else:
                X_resampled_list.append(X_cls)
                y_resampled_list.append(y_cls)
            del X_cls, y_cls
            gc.collect()
        X_balanced = np.vstack(X_resampled_list)
        y_balanced = np.concatenate(y_resampled_list)
        X_final = self.standard_scaler.inverse_transform(X_balanced)
        X_final = X_final / feature_importance
        del X_balanced, X_scaled, X_weighted
        gc.collect()
        return X_final, y_balanced

if len(X_full) > 300000:
    sample_idx = np.random.choice(len(X_full), 300000, replace=False)
    X_full = X_full[sample_idx]
    y_full = y_full[sample_idx]

sampler = OptimizedNeuroSyncSampler(
    neural_depth=2,
    evolutionary_generations=3,
    target_samples_per_class=5000,
    random_state=42
)
X_resampled, y_resampled = sampler.fit_resample(X_full, y_full)

df_balanced = pd.DataFrame(X_resampled, columns=available_features)
df_balanced['Label'] = y_resampled

df_balanced.to_csv('fraud_balanced.csv', index=False)
del X_full, y_full, X_resampled, y_resampled
gc.collect()

df = pd.read_csv('fraud_balanced.csv')

drop_cols = ['TransactionID', 'isFraud']
df.drop(columns=[c for c in drop_cols if c in df.columns], inplace=True, errors='ignore')
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

constant_cols = df.columns[df.nunique() <= 1]
df.drop(columns=constant_cols, inplace=True)

le = LabelEncoder()
df['Label_encoded'] = le.fit_transform(df['Label'])
label_map = dict(zip(le.classes_, le.transform(le.classes_)))

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
if 'Label_encoded' in numeric_cols:
    numeric_cols.remove('Label_encoded')

X = df[numeric_cols].values
y = to_categorical(df['Label_encoded'].values)

X = X.reshape((X.shape[0], 1, X.shape[1]))
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train.reshape(X_train.shape[0], -1)).reshape(X_train.shape)
X_test_scaled = scaler.transform(X_test.reshape(X_test.shape[0], -1)).reshape(X_test.shape)

def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0):
    x = LayerNormalization(epsilon=1e-6)(inputs)
    x = MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(x, x)
    x = Add()([inputs, x])
    res = x
    x = LayerNormalization(epsilon=1e-6)(x)
    x = Dense(ff_dim, activation="relu")(x)
    x = Dropout(dropout)(x)
    x = Dense(inputs.shape[-1])(x)
    x = Add()([res, x])
    return x

@tf.custom_gradient
def gradient_reverse_op(x, lambda_):
    def grad(dy):
        return -lambda_ * dy, None
    return x, grad

class GradientReversalLayer(tf.keras.layers.Layer):
    def __init__(self, lambda_=1.0, **kwargs):
        super(GradientReversalLayer, self).__init__(**kwargs)
        self.lambda_ = lambda_

    def call(self, x, training=None):
        if training:
            return gradient_reverse_op(x, self.lambda_)
        return x

    def compute_output_shape(self, input_shape):
        return input_shape

    def get_config(self):
        config = super().get_config()
        config.update({'lambda_': self.lambda_})
        return config

def build_cnn_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Conv1D(filters=32, kernel_size=3, activation='relu', padding='same')(permuted)
    x = MaxPooling1D(pool_size=2)(x)
    x = Conv1D(filters=64, kernel_size=3, activation='relu', padding='same')(x)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_bigru_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Bidirectional(GRU(64, return_sequences=True))(permuted)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_transformer_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)
    x = Dense(64, activation='relu')(permuted)
    x = transformer_encoder(x, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    x = GlobalAveragePooling1D()(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_proposed_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)

    cnn = Conv1D(32, 3, activation='relu', padding='same')(permuted)
    cnn = MaxPooling1D(2)(cnn)
    cnn = Conv1D(64, 3, activation='relu', padding='same')(cnn)
    cnn = GlobalAveragePooling1D()(cnn)
    cnn = Dense(64, activation='relu')(cnn)

    gru = Bidirectional(GRU(64, return_sequences=True))(permuted)
    gru = GlobalAveragePooling1D()(gru)
    gru = Dense(64, activation='relu')(gru)

    trans = Dense(64, activation='relu')(permuted)
    trans = transformer_encoder(trans, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    trans = GlobalAveragePooling1D()(trans)
    trans = Dense(64, activation='relu')(trans)

    fused = Lambda(lambda t: tf.stack(t, axis=1), output_shape=(3, 64))([cnn, gru, trans])
    attn = MultiHeadAttention(num_heads=4, key_dim=32)(fused, fused)
    attn = Flatten()(attn)
    x = Dense(64, activation='relu')(attn)
    x = Dropout(0.5)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
    return model

def build_adversarial_game_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    permuted = Permute((2, 1))(inputs)

    cnn_branch = Conv1D(filters=32, kernel_size=3, activation='relu', padding='same')(permuted)
    cnn_branch = MaxPooling1D(pool_size=2)(cnn_branch)
    cnn_branch = Conv1D(filters=64, kernel_size=3, activation='relu', padding='same')(cnn_branch)
    cnn_branch = GlobalAveragePooling1D()(cnn_branch)
    cnn_branch = Dense(64, activation='relu')(cnn_branch)

    gru_branch = Bidirectional(GRU(64, return_sequences=True))(permuted)
    gru_branch = GlobalAveragePooling1D()(gru_branch)
    gru_branch = Dense(64, activation='relu')(gru_branch)

    transformer_proj = Dense(64, activation='relu')(permuted)
    transformer_branch = transformer_encoder(transformer_proj, head_size=32, num_heads=4, ff_dim=64, dropout=0.2)
    transformer_branch = GlobalAveragePooling1D()(transformer_branch)
    transformer_branch = Dense(64, activation='relu')(transformer_branch)

    fused = Lambda(lambda tensors: tf.stack(tensors, axis=1),
                   output_shape=(3, 64))([cnn_branch, gru_branch, transformer_branch])
    attention = MultiHeadAttention(num_heads=4, key_dim=32)(fused, fused)
    attention = Flatten()(attention)
    fusion_features = Dense(64, activation='relu', name='fusion_features')(attention)

    x = Dropout(0.5)(fusion_features)
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.3)(x)
    class_outputs = Dense(num_classes, activation='softmax', name='classifier')(x)

    adv_branch = GradientReversalLayer(lambda_=0.1, name='grad_reverse')(fusion_features)
    adv_branch = Dense(32, activation='relu')(adv_branch)
    adv_branch = Dense(32, activation='relu')(adv_branch)
    disc_outputs = Dense(1, activation='sigmoid', name='discriminator')(adv_branch)

    model = Model(inputs, [class_outputs, disc_outputs])
    return model

class AdversarialGameModel(Model):
    def __init__(self, base_model, epsilon=0.01, **kwargs):
        super(AdversarialGameModel, self).__init__(**kwargs)
        self.base_model = base_model
        self.epsilon = epsilon
        self.disc_loss_tracker = tf.keras.metrics.Mean(name='disc_loss')
        self.class_loss_tracker = tf.keras.metrics.Mean(name='class_loss')
        self.total_loss_tracker = tf.keras.metrics.Mean(name='loss')
        self.train_acc_tracker = tf.keras.metrics.CategoricalAccuracy(name='train_accuracy')
        self.val_acc_tracker = tf.keras.metrics.CategoricalAccuracy(name='val_accuracy')

    @property
    def metrics(self):
        return [
            self.disc_loss_tracker,
            self.class_loss_tracker,
            self.total_loss_tracker,
            self.train_acc_tracker,
            self.val_acc_tracker,
        ]

    def compile(self, optimizer, class_loss_fn, disc_loss_fn):
        super(AdversarialGameModel, self).compile(optimizer=optimizer)
        self.class_loss_fn = class_loss_fn
        self.disc_loss_fn = disc_loss_fn

    def generate_adversarial(self, x, y_true):
        x = tf.cast(x, tf.float32)
        with tf.GradientTape() as tape:
            tape.watch(x)
            preds, _ = self.base_model(x, training=True)
            loss = self.class_loss_fn(y_true, preds)
        grad = tape.gradient(loss, x)
        if grad is None:
            return x
        signed_grad = tf.sign(grad)
        x_adv = x + self.epsilon * signed_grad
        x_adv = tf.clip_by_value(x_adv, tf.reduce_min(x), tf.reduce_max(x))
        return x_adv

    def train_step(self, data):
        x, y_true = data
        batch_size = tf.shape(x)[0]

        x_adv = self.generate_adversarial(x, y_true)
        clean_labels = tf.ones((batch_size, 1))
        adv_labels = tf.zeros((batch_size, 1))
        combined_x = tf.concat([x, x_adv], axis=0)
        combined_disc_labels = tf.concat([clean_labels, adv_labels], axis=0)

        with tf.GradientTape() as tape:
            class_preds, disc_preds = self.base_model(combined_x, training=True)
            class_preds_clean = class_preds[:batch_size]
            class_preds_adv = class_preds[batch_size:]
            class_loss_clean = self.class_loss_fn(y_true, class_preds_clean)
            class_loss_adv = self.class_loss_fn(y_true, class_preds_adv)
            class_loss = class_loss_clean + 0.5 * class_loss_adv
            disc_loss = self.disc_loss_fn(combined_disc_labels, disc_preds)
            total_loss = class_loss + disc_loss

        trainable_vars = self.base_model.trainable_variables
        grads = tape.gradient(total_loss, trainable_vars)
        self.optimizer.apply_gradients(zip(grads, trainable_vars))

        self.total_loss_tracker.update_state(total_loss)
        self.class_loss_tracker.update_state(class_loss)
        self.disc_loss_tracker.update_state(disc_loss)
        self.train_acc_tracker.update_state(y_true, class_preds_clean)

        return {
            'loss': self.total_loss_tracker.result(),
            'class_loss': self.class_loss_tracker.result(),
            'disc_loss': self.disc_loss_tracker.result(),
            'accuracy': self.train_acc_tracker.result(),
        }

    def test_step(self, data):
        x, y_true = data
        class_preds, disc_preds = self.base_model(x, training=False)
        class_loss = self.class_loss_fn(y_true, class_preds)
        clean_labels = tf.ones((tf.shape(x)[0], 1))
        disc_loss = self.disc_loss_fn(clean_labels, disc_preds)
        total_loss = class_loss + disc_loss

        self.total_loss_tracker.update_state(total_loss)
        self.class_loss_tracker.update_state(class_loss)
        self.disc_loss_tracker.update_state(disc_loss)
        self.val_acc_tracker.update_state(y_true, class_preds)

        return {
            'loss': self.total_loss_tracker.result(),
            'class_loss': self.class_loss_tracker.result(),
            'disc_loss': self.disc_loss_tracker.result(),
            'accuracy': self.val_acc_tracker.result(),
        }

    def call(self, inputs, training=False):
        class_preds, _ = self.base_model(inputs, training=training)
        return class_preds

class StopAtAccuracy(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        if logs.get('val_accuracy') is not None and logs.get('val_accuracy') >= 0.98:
            self.model.stop_training = True
            print(f"\nReached 98% validation accuracy, stopping training at epoch {epoch+1}")

def train_and_evaluate(model, model_name, X_train, y_train, X_test, y_test, target_names):
    early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
    stop_acc = StopAtAccuracy()
    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=20,
        batch_size=64,
        callbacks=[early_stop, stop_acc],
        verbose=1
    )
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title(f'{model_name} Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend(['Train', 'Val'], loc='upper right')
    plt.show()
    plt.plot(history.history['loss'])
    plt.plot(history.history['val_loss'])
    plt.title(f'{model_name} Loss')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend(['Train', 'Val'], loc='upper right')
    plt.show()
    y_pred_probs = model.predict(X_test)
    y_pred = np.argmax(y_pred_probs, axis=1)
    y_true = np.argmax(y_test, axis=1)
    print(f"\n{model_name} Classification Report:")
    report = classification_report(y_true, y_pred, target_names=target_names, zero_division=0, output_dict=True)
    print(classification_report(y_true, y_pred, target_names=target_names, zero_division=0))
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=target_names, yticklabels=target_names)
    plt.title(f'{model_name} Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.show()
    plt.figure(figsize=(8, 6))
    for i in range(len(target_names)):
        fpr, tpr, _ = roc_curve(y_test[:, i], y_pred_probs[:, i])
        plt.plot(fpr, tpr, label=f'{target_names[i]} (AUC = {auc(fpr, tpr):.2f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{model_name} ROC Curves')
    plt.legend(loc='lower right')
    plt.show()
    auc_score = roc_auc_score(y_test, y_pred_probs, multi_class="ovr")
    print(f"{model_name} AUC Score (OvR): {auc_score:.4f}")
    return report, auc_score

target_names = [str(c) for c in le.classes_]

cnn_model = build_cnn_model(X_train_scaled.shape[1:], y_train.shape[1])
bigru_model = build_bigru_model(X_train_scaled.shape[1:], y_train.shape[1])
transformer_model = build_transformer_model(X_train_scaled.shape[1:], y_train.shape[1])
proposed_model = build_proposed_model(X_train_scaled.shape[1:], y_train.shape[1])

base_proposed_model = build_adversarial_game_model(X_train_scaled.shape[1:], y_train.shape[1])
adversarial_game_model = AdversarialGameModel(base_proposed_model, epsilon=0.01)
adversarial_game_model.compile(
    optimizer=Adam(learning_rate=0.0005),
    class_loss_fn=tf.keras.losses.CategoricalCrossentropy(),
    disc_loss_fn=tf.keras.losses.BinaryCrossentropy()
)

print("Training CNN Model")
cnn_report, cnn_auc = train_and_evaluate(cnn_model, "CNN", X_train_scaled, y_train, X_test_scaled, y_test, target_names)

print("Training BiGRU Model")
bigru_report, bigru_auc = train_and_evaluate(bigru_model, "BiGRU", X_train_scaled, y_train, X_test_scaled, y_test, target_names)

print("Training Transformer Model")
transformer_report, transformer_auc = train_and_evaluate(transformer_model, "Transformer", X_train_scaled, y_train, X_test_scaled, y_test, target_names)

print("Training Proposed Hybrid Fusion Model")
proposed_report, proposed_auc = train_and_evaluate(proposed_model, "Proposed", X_train_scaled, y_train, X_test_scaled, y_test, target_names)

print("Training Proposed Hybrid Fusion Model with Adversarial Game Theory")
y_train_single = y_train
y_test_single = y_test

adversarial_game_model.fit(
    X_train_scaled, y_train_single,
    validation_data=(X_test_scaled, y_test_single),
    epochs=20,
    batch_size=64,
    callbacks=[EarlyStopping(monitor='val_loss', mode='min', patience=5, restore_best_weights=True), StopAtAccuracy()],
    verbose=1
)

plt.plot(adversarial_game_model.history.history['accuracy'])
plt.plot(adversarial_game_model.history.history['val_accuracy'])
plt.title('Proposed + Adversarial Game Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Val'], loc='upper right')
plt.show()

plt.plot(adversarial_game_model.history.history['loss'])
plt.plot(adversarial_game_model.history.history['val_loss'])
plt.title('Proposed + Adversarial Game Total Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Val'], loc='upper right')
plt.show()

plt.plot(adversarial_game_model.history.history['class_loss'])
plt.title('Proposed + Adversarial Game Classifier Loss')
plt.ylabel('Class Loss')
plt.xlabel('Epoch')
plt.show()

plt.plot(adversarial_game_model.history.history['disc_loss'])
plt.title('Proposed + Adversarial Game Discriminator Loss')
plt.ylabel('Disc Loss')
plt.xlabel('Epoch')
plt.show()

class_preds_proposed = adversarial_game_model.predict(X_test_scaled)
y_pred_proposed = np.argmax(class_preds_proposed, axis=1)
y_true_proposed = np.argmax(y_test_single, axis=1)

print("\nProposed + Adversarial Game Classification Report:")
adver_game_report = classification_report(y_true_proposed, y_pred_proposed, target_names=target_names, zero_division=0, output_dict=True)
print(classification_report(y_true_proposed, y_pred_proposed, target_names=target_names, zero_division=0))

cm_proposed = confusion_matrix(y_true_proposed, y_pred_proposed)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_proposed, annot=True, fmt="d", cmap="Blues", xticklabels=target_names, yticklabels=target_names)
plt.title('Proposed + Adversarial Game Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()

cm_proposed_norm = confusion_matrix(y_true_proposed, y_pred_proposed, normalize='true')
plt.figure(figsize=(8, 6))
sns.heatmap(cm_proposed_norm, annot=True, fmt=".2f", cmap="Blues", xticklabels=target_names, yticklabels=target_names)
plt.title('Proposed + Adversarial Game Confusion Matrix (Normalized)')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()

plt.figure(figsize=(8, 6))
for i in range(len(target_names)):
    fpr, tpr, _ = roc_curve(y_test_single[:, i], class_preds_proposed[:, i])
    plt.plot(fpr, tpr, label=f'{target_names[i]} (AUC = {auc(fpr, tpr):.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Proposed + Adversarial Game ROC Curves')
plt.legend(loc='lower right')
plt.show()

adver_game_auc = roc_auc_score(y_test_single, class_preds_proposed, multi_class="ovr")
print(f"Proposed + Adversarial Game AUC Score (OvR): {adver_game_auc:.4f}")

print("\nAdversarial Game Specific Evaluation:")
X_test_tf = tf.cast(X_test_scaled, tf.float32)
y_test_tf = tf.cast(y_test_single, tf.float32)

with tf.GradientTape() as tape:
    tape.watch(X_test_tf)
    preds, _ = adversarial_game_model.base_model(X_test_tf, training=False)
    loss = tf.keras.losses.CategoricalCrossentropy()(y_test_tf, preds)
grad = tape.gradient(loss, X_test_tf)
X_test_adv = X_test_tf + 0.01 * tf.sign(grad)
X_test_adv = tf.clip_by_value(X_test_adv, tf.reduce_min(X_test_tf), tf.reduce_max(X_test_tf))

_, disc_clean = adversarial_game_model.base_model(X_test_tf, training=False)
_, disc_adv = adversarial_game_model.base_model(X_test_adv, training=False)
disc_clean = disc_clean.numpy().ravel()
disc_adv = disc_adv.numpy().ravel()

disc_labels = np.concatenate([np.ones(len(disc_clean)), np.zeros(len(disc_adv))])
disc_scores = np.concatenate([disc_clean, disc_adv])
disc_auc = roc_auc_score(disc_labels, disc_scores)
disc_acc = np.mean((disc_scores >= 0.5).astype(int) == disc_labels)

print(f"Discriminator Accuracy (clean vs adversarial): {disc_acc:.4f}")
print(f"Discriminator AUC:                             {disc_auc:.4f}")
print(f"Mean disc score on CLEAN samples:              {disc_clean.mean():.4f}")
print(f"Mean disc score on ADVERSARIAL samples:        {disc_adv.mean():.4f}")

fpr_d, tpr_d, _ = roc_curve(disc_labels, disc_scores)
plt.figure(figsize=(8, 6))
plt.plot(fpr_d, tpr_d, label=f'Discriminator (AUC = {disc_auc:.3f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Proposed + Adversarial Game Discriminator ROC')
plt.legend(loc='lower right')
plt.show()

adv_preds = adversarial_game_model.predict(X_test_adv.numpy())
clean_acc = np.mean(y_pred_proposed == y_true_proposed)
adv_acc = np.mean(np.argmax(adv_preds, axis=1) == y_true_proposed)
print(f"Classifier accuracy on CLEAN test:       {clean_acc:.4f}")
print(f"Classifier accuracy on ADVERSARIAL test: {adv_acc:.4f}")
print(f"Robustness drop (clean - adversarial):   {clean_acc - adv_acc:.4f}")

models = ['CNN', 'BiGRU', 'Transformer', 'Proposed', 'Prop+AdvGame']
accuracy = [cnn_report['accuracy'], bigru_report['accuracy'], transformer_report['accuracy'], proposed_report['accuracy'], adver_game_report['accuracy']]
precision = [cnn_report['macro avg']['precision'], bigru_report['macro avg']['precision'], transformer_report['macro avg']['precision'], proposed_report['macro avg']['precision'], adver_game_report['macro avg']['precision']]
recall = [cnn_report['macro avg']['recall'], bigru_report['macro avg']['recall'], transformer_report['macro avg']['recall'], proposed_report['macro avg']['recall'], adver_game_report['macro avg']['recall']]
f1 = [cnn_report['macro avg']['f1-score'], bigru_report['macro avg']['f1-score'], transformer_report['macro avg']['f1-score'], proposed_report['macro avg']['f1-score'], adver_game_report['macro avg']['f1-score']]
auc_scores = [cnn_auc, bigru_auc, transformer_auc, proposed_auc, adver_game_auc]

x = np.arange(len(models))
width = 0.15
fig, ax = plt.subplots(figsize=(14, 6))
bar1 = ax.bar(x - 2*width, accuracy, width, label='Accuracy')
bar2 = ax.bar(x - width, precision, width, label='Precision')
bar3 = ax.bar(x, recall, width, label='Recall')
bar4 = ax.bar(x + width, f1, width, label='F1-score')
bar5 = ax.bar(x + 2*width, auc_scores, width, label='AUC')
ax.set_xlabel('Models')
ax.set_ylabel('Scores')
ax.set_title('Model Performance Comparison (IEEE Fraud Detection)')
ax.set_xticks(x)
ax.set_xticklabels(models)
ax.legend()
plt.tight_layout()
plt.show()

summary_df = pd.DataFrame({
    'Model': models,
    'Accuracy': accuracy,
    'Precision': precision,
    'Recall': recall,
    'F1-Score': f1,
    'AUC (OvR)': auc_scores,
})
print("\n=== Overall Model Comparison Table ===")
print(summary_df.round(4).to_string(index=False))

print("\n=== Adversarial Game Summary (Proposed + Adversarial Game) ===")
print(f"Clean Test Accuracy:             {clean_acc:.4f}")
print(f"Adversarial Test Accuracy:       {adv_acc:.4f}")
print(f"Robustness Drop (clean - adv):   {clean_acc - adv_acc:.4f}")
print(f"Discriminator AUC:               {disc_auc:.4f}")
```
